# Data Cleaning with Pandas

```python
import pandas as pd
```

```python
df=pd.read_excel(r"D:\CHITRA 8-3-2026\Data_Projects\Pandas_Project\Data_Cleaning_Pandas1\customer_call_list.xlsx")
df  #reading the file
```

```python
df.drop_duplicates() # Removes duplicates from the file
```

```python
df.drop(columns ="Not_Useful_Column") #Remove the Unwanted Column
```

```python
df["Last_Name"]=df["Last_Name"].str.strip("./_")  #Removingt the unwanted characters from the colm
df
```

```python
df["Phone_Number"]=df["Phone_Number"].str.replace('[^a-zA-Z0-9]','',regex=True) #Removing unwanted symbols and placing with nothing
```

```python
df["Phone_Number"]=df["Phone_Number"].apply(lambda x: str(x))
df#changing the phone numbers into string
```

```python
df["Phone_Number"]=df["Phone_Number"].apply(lambda x: x[0:3]+'-'+ x[3:6] +'-'+ x[6:10])  #Changing the formate of the Number
df
```

```python
df["Phone_Number"]=df["Phone_Number"].str.replace('nan--','') # Removing 'nan--' from the colm
df["Phone_Number"]=df["Phone_Number"].str.replace('Na--','')  # Removing 'Na--' from the colm
df
```

```python
df[["Street_Address","City","Zip_Code"]]=df["Address"].str.split(',',n=2,expand=True) #Spliting the Address colm category wise
df
```

```python
df.drop(columns='Address')
```

```python
df["Paying Customer"]=df["Paying Customer"].replace({'Y':'Yes'}) #Replacing all the Y's into Yes with the exact match technique.
df["Paying Customer"]=df["Paying Customer"].replace({'N':'No'}) #Replacing all the N's into No with the exact match technique.
```

```python
df
```

```python
df["Do_Not_Contact"]=df["Do_Not_Contact"].replace({'Y':'Yes'}) #Replacing all the Y's into Yes with the exact match technique.
df["Do_Not_Contact"]=df["Do_Not_Contact"].replace({'N':'No'})#Replacing all the N's into No with the exact match technique.
df
```

```python
df=df.fillna('')
df
```

```python
df=df.replace('N/a','')
df
```

# Requirement: Only details of the Customers who are with representative calling them.

```python
df=df[df["Do_Not_Contact"] != 'Yes'] # Keeping only the Cutomers ok with rep. calling them
df
```

```python
#Cleans the hidden space in the column name and filters out true missing values
#df = df[df["Phone_Number"].notna()]
        #or
df=df[df["Phone_Number"] != ''] #Removing all the Blank colm without contact
df
```
